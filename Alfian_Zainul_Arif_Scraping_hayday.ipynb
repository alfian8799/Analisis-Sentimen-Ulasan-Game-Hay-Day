# **Scraping Data**

```python
# scraping_data.py
from google_play_scraper import reviews, Sort
import pandas as pd

def scrape_playstore_data():
    print("Memulai proses scraping data ulasan game Hay Day...")

    # Mengambil ulasan (Target 12.000 ulasan untuk memenuhi syarat > 10.000 sampel)
    result, continuation_token = reviews(
        'com.supercell.hayday',
        lang='id',             # Bahasa ulasan: Indonesia
        country='id',          # Negara: Indonesia
        sort=Sort.NEWEST,      # Mengambil ulasan paling baru
        count=12000            # Jumlah data
    )

    # Mengubah hasil ke dalam bentuk DataFrame Pandas
    df = pd.DataFrame(result)
    print(f"Total data mentah yang berhasil diambil: {df.shape[0]} baris")

    # Menyimpan hasil scraping ke CSV
    nama_file = 'ulasan_hayday_mentah.csv'
    df.to_csv(nama_file, index=False)
    print(f"Data berhasil disimpan ke dalam file: {nama_file}")

if __name__ == "__main__":
    scrape_playstore_data()
```
